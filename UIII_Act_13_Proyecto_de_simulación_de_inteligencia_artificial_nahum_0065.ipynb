```python
# ==========================================================
# SISTEMA IA - BURÓ DE CRÉDITO
# CBTIS 128
# Carrera: Inteligencia Artificial
# Materia: Elabora Proyectos de Programación Lógica
# Docente: Eliseo Nava
# Alumno: Nahum Flores
# NL = 21
# ==========================================================

class Buro_De_Credito:

    # Constructor
    def __init__(self):
        self.lista_clientes = []

    # ======================================================
    # FUNCIÓN: AGREGAR CLIENTE
    # ======================================================

    def agregar_cliente(self):

        print("\n===== REGISTRO DE CLIENTE =====")

        nombre = input("Nombre del cliente: ")
        numero_credito = input("Número de crédito: ")
        historial = input("Historial de pagos: ")
        deuda = input("Deuda actual: ")
        riesgo = input("Nivel de riesgo: ")

        cliente = {
            "nombre": nombre,
            "numero_credito": numero_credito,
            "historial": historial,
            "deuda": deuda,
            "riesgo": riesgo
        }

        self.lista_clientes.append(cliente)

        print("\n✅ Cliente registrado correctamente.")

    # ======================================================
    # FUNCIÓN: LISTAR CLIENTES
    # ======================================================

    def listar_clientes(self):

        print("\n================ LISTA DE CLIENTES =================")

        if len(self.lista_clientes) == 0:
            print("No hay clientes registrados.")

        else:

            print(
                f"{'No.':<5}"
                f"{'NOMBRE':<20}"
                f"{'CRÉDITO':<18}"
                f"{'HISTORIAL':<20}"
                f"{'DEUDA':<15}"
                f"{'RIESGO':<15}"
            )

            print("-" * 93)

            contador = 1

            for cliente in self.lista_clientes:

                print(
                    f"{contador:<5}"
                    f"{cliente['nombre']:<20}"
                    f"{cliente['numero_credito']:<18}"
                    f"{cliente['historial']:<20}"
                    f"{cliente['deuda']:<15}"
                    f"{cliente['riesgo']:<15}"
                )

                contador += 1

    # ======================================================
    # FUNCIÓN: ACTUALIZAR CLIENTE
    # ======================================================

    def actualizar_cliente(self):

        self.listar_clientes()

        if len(self.lista_clientes) > 0:

            numero = int(input("\nNúmero del cliente a actualizar: "))

            if numero >= 1 and numero <= len(self.lista_clientes):

                print("\n===== NUEVOS DATOS =====")

                self.lista_clientes[numero - 1]["nombre"] = input("Nuevo nombre: ")
                self.lista_clientes[numero - 1]["numero_credito"] = input("Nuevo número de crédito: ")
                self.lista_clientes[numero - 1]["historial"] = input("Nuevo historial: ")
                self.lista_clientes[numero - 1]["deuda"] = input("Nueva deuda: ")
                self.lista_clientes[numero - 1]["riesgo"] = input("Nuevo nivel de riesgo: ")

                print("\n✅ Cliente actualizado correctamente.")

            else:
                print("❌ Número incorrecto.")

    # ======================================================
    # FUNCIÓN: BORRAR CLIENTE
    # ======================================================

    def borrar_cliente(self):

        self.listar_clientes()

        if len(self.lista_clientes) > 0:

            numero = int(input("\nNúmero del cliente a borrar: "))

            if numero >= 1 and numero <= len(self.lista_clientes):

                self.lista_clientes.pop(numero - 1)

                print("\n🗑️ Cliente eliminado correctamente.")

            else:
                print("❌ Número incorrecto.")

    # ======================================================
    # FUNCIÓN: CONSULTAR RIESGO
    # ======================================================

    def consultar_riesgo(self):

        self.listar_clientes()

        if len(self.lista_clientes) > 0:

            numero = int(input("\nNúmero del cliente a consultar: "))

            if numero >= 1 and numero <= len(self.lista_clientes):

                cliente = self.lista_clientes[numero - 1]

                print("\n===== CONSULTA DE RIESGO =====")
                print("Cliente:", cliente["nombre"])
                print("Nivel de riesgo:", cliente["riesgo"])

            else:
                print("❌ Número incorrecto.")


# ==========================================================
# PROGRAMA PRINCIPAL
# ==========================================================

sistema = Buro_De_Credito()

opcion = 0

while opcion != 6:

    print("\n======================================")
    print("     IA DE BURÓ DE CRÉDITO")
    print("======================================")
    print("1. Agregar cliente")
    print("2. Listar clientes")
    print("3. Actualizar cliente")
    print("4. Borrar cliente")
    print("5. Consultar riesgo")
    print("6. Salir")
    print("======================================")

    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
        sistema.agregar_cliente()

    elif opcion == 2:
        sistema.listar_clientes()

    elif opcion == 3:
        sistema.actualizar_cliente()

    elif opcion == 4:
        sistema.borrar_cliente()

    elif opcion == 5:
        sistema.consultar_riesgo()

    elif opcion == 6:
        print("\n👋 Saliendo del sistema...")
        print("Gracias por utilizar el sistema IA.")

    else:
        print("\n❌ Opción inválida.")
```
